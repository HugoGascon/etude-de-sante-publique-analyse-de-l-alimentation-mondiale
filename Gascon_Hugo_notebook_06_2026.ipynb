<h3 style="margin: auto; padding: 20px; color: RGB(0,150,250); ">3.3 - Nombre théorique de personne qui pourrait être nourrie avec les produits végétaux</h3>
</div>

```python
#Transfert des données avec les végétaux dans un nouveau dataframe
dispo_vegetaux = dispo_alimentaire[dispo_alimentaire['Origine'] == 'vegetale'].copy()
print(dispo_vegetaux.head(5))
```

```python
#Calcul du nombre de kcal disponible pour les végétaux
# 1. Vérifier si la colonne existe (débogage)
print("Colonnes disponibles dans dispo_vegetaux :", dispo_vegetaux.columns.tolist())

# 2. Si elle n'existe pas, il faut refaire la jointure avec le dataframe 'population'
# Assure-toi que 'population' est bien le dataframe filtré sur l'année 2017 ou l'année voulue
dispo_vegetaux = pd.merge(
    dispo_vegetaux,
    population[['Zone', 'Population']], # On prend uniquement la colonne nécessaire
    on='Zone',
    how='left'
)

# 3. Maintenant, tu peux refaire ton calcul
dispo_vegetaux['dispo_kcal_totale'] = (
    dispo_vegetaux['Disponibilité alimentaire (Kcal/personne/jour)'] * 365 * dispo_vegetaux['Population']
)
print(dispo_vegetaux.head(5))
```

```python
#Calcul du nombre d'humains pouvant être nourris avec les végétaux
besoin_annuel_kcal_humain = 2500 * 365
nb_humains_nourris = dispo_vegetaux['dispo_kcal_totale'].sum() / besoin_annuel_kcal_humain
print(f"Nombre d'humains pouvant être nourris avec les végétaux : {nb_humains_nourris:.2f}")
```

```python
#Calcul de la disponibilité totale
disponibilite_interieure_somme = dispo_alimentaire["Disponibilité intérieure"].sum()
print(f"Disponibilité intérieure totale : {disponibilite_interieure_somme:.2f} kg")
```

```python
#création d'une boucle for pour afficher les différentes valeurs en fonction des colonnes aliments pour animaux, pertes, nourritures, 

colonnes_boucle_for = ["Aliments pour animaux", "Autres Utilisations", "Nourriture", "Pertes", "Semences", "Traitement"]


for elt in colonnes_boucle_for:
  
    prop_elt_totale = dispo_alimentaire[elt].sum()
    
   
    prop_elt_pourcentage = (prop_elt_totale / disponibilite_interieure_somme) * 100
    
    
    print(f"Total {elt:25} : {prop_elt_totale:15,.0f} kg ({prop_elt_pourcentage:5.2f}%)")
```

<h3 style="margin: auto; padding: 20px; color: RGB(0,150,250); ">3.7 - Pays qui ont le plus bénéficié d'aide alimentaire depuis 2013</h3>
</div>

```python
#calcul du total de l'aide alimentaire par pays
# Groupement par pays (Zone) et somme de l'aide alimentaire reçue
aide_totale_par_pays = aide_alimentaire.groupby('Zone')['Valeur'].sum().reset_index()

# Pour plus de clarté, on renomme la colonne 'Valeur' par 'Aide_totale_kg'
aide_totale_par_pays.rename(columns={'Valeur': 'Aide_totale_kg'}, inplace=True)

print("Le calcul du cumul de l'aide alimentaire par pays a été réalisé avec succès !")
```

```python
#affichage après trie des 10 pays qui ont bénéficié le plus de l'aide alimentaire
# Tri par ordre décroissant de l'aide totale et sélection des 10 premiers pays
top10_aide = aide_totale_par_pays.sort_values(by='Aide_totale_kg', ascending=False).head(10)

# Affichage propre avec conversion en tonnes pour que la lecture soit plus digeste
print("=== TOP 10 DES PAYS AYANT LE PLUS BÉNÉFICIÉ D'AIDE ALIMENTAIRE DEPUIS 2013 ===")
for index, row in top10_aide.iterrows():
    # Division par 1000 pour réafficher en Tonnes lors de la présentation (plus lisible pour des volumes si grands !)
    aide_en_tonnes = row['Aide_totale_kg'] / 1000
    print(f"{row['Zone']:30} : {aide_en_tonnes:15,.0f} tonnes")
```

<div style="border: 1px solid RGB(0,150,250);" >
<h3 style="margin: auto; padding: 20px; color: RGB(0,150,250); ">3.8 - Evolution des 5 pays qui ont le plus bénéficiés de l'aide alimentaire entre 2013 et 2016</h3>
</div>

```python
#Création d'un dataframe avec la zone, l'année et l'aide alimentaire puis groupby sur zone et année 
# 1. Identifier les 5 pays ayant reçu le plus d'aide alimentaire
top5_pays = aide_totale_par_pays.sort_values(by='Aide_totale_kg', ascending=False).head(5)['Zone'].tolist()

# 2. Filtrer le dataframe principal d'aide alimentaire pour ne garder que ces 5 pays
pays_beneficiaires_aide_alimentaire = aide_alimentaire[aide_alimentaire['Zone'].isin(top5_pays)].copy()

# 3. Grouper par pays (Zone) et par Année pour avoir le détail annuel
evolution_aide_top5 = pays_beneficiaires_aide_alimentaire.groupby(['Zone', 'Année'])['Valeur'].sum().reset_index()

# Renommer la colonne par clarté (on convertit en tonnes pour la lisibilité)
evolution_aide_top5.rename(columns={'Valeur': 'Aide_annuelle_kg'}, inplace=True)
evolution_aide_top5['Aide_annuelle_tonnes'] = evolution_aide_top5['Aide_annuelle_kg'] / 1000

print("Dataframe d'évolution créé pour les 5 pays suivants :", top5_pays)
```

```python
#Création d'une liste contenant les 5 pays qui ont le plus bénéficiées de l'aide alimentaire
# Sélection des 5 pays ayant reçu le plus d'aide alimentaire et conversion en liste
top5_pays_beneficiaires_aide_alimentaire = aide_totale_par_pays.sort_values(by='Aide_totale_kg', ascending=False).head(5)['Zone'].tolist()

# Affichage de la liste pour vérification
print("Liste des 5 pays :", top5_pays_beneficiaires_aide_alimentaire)
```

```python
#On filtre sur le dataframe avec notre liste
# Sélection des 5 pays ayant reçu le plus d'aide alimentaire et conversion en liste
top5_pays_beneficiaires_aide_alimentaire = aide_totale_par_pays.sort_values(by='Aide_totale_kg', ascending=False).head(5)['Zone'].tolist()

# Affichage de la liste pour vérification
print("Liste des 5 pays :", top5_pays_beneficiaires_aide_alimentaire)
```

```python
# Affichage des pays avec l'aide alimentaire par année
for pays in top5_pays_beneficiaires_aide_alimentaire:
    print(f"\n=== ÉVOLUTION DE L'AIDE ALIMENTAIRE POUR {pays.upper()} ===")
    aide_par_annee = evolution_aide_top5[evolution_aide_top5['Zone'] == pays]
    for index, row in aide_par_annee.iterrows():
        print(f"Année {row['Année']}: {row['Aide_annuelle_tonnes']:,.0f} tonnes")
```

```python
# Somme des protéines par pays et par origine (animale vs végétale)
dispo_proteines = dispo_alimentaire.groupby(['Zone', 'Origine'])['Disponibilité de protéines en quantité (g/personne/jour)'].sum().unstack().fillna(0)

# Calcul de la part de protéines végétales
dispo_proteines['Part_Vegetale_%'] = (dispo_proteines['vegetale'] / (dispo_proteines['animale'] + dispo_proteines['vegetale'])) * 100

# Top 10 des pays avec la plus faible dispo en protéines totales
dispo_proteines['Total_Proteines'] = dispo_proteines['animale'] + dispo_proteines['vegetale']
print(dispo_proteines.sort_values(by='Total_Proteines').head(10))
```

# Analyse critique des "Pertes" (Le gaspillage alimentaire)

```python
# Taux de perte par produit au niveau mondial
pertes_par_produit = dispo_alimentaire.groupby('Produit')[['Pertes', 'Disponibilité intérieure']].sum()
pertes_par_produit['Taux_de_perte_%'] = (pertes_par_produit['Pertes'] / pertes_par_produit['Disponibilité intérieure']) * 100

# Afficher les produits avec plus de 100 000 tonnes de dispo qui ont le plus fort taux de perte
print(pertes_par_produit[pertes_par_produit['Disponibilité intérieure'] > 100000].sort_values(by='Taux_de_perte_%', ascending=False).head(10))
```

```python
import matplotlib.pyplot as plt

# 1. Identifier les 5 pays et filtrer
top5_pays = aide_totale_par_pays.sort_values(by='Aide_totale_kg', ascending=False).head(5)['Zone'].tolist()
evolution_top5 = aide_alimentaire[aide_alimentaire['Zone'].isin(top5_pays)].groupby(['Zone', 'Année'])['Valeur'].sum().reset_index()

# 2. Création du graphique
plt.figure(figsize=(10, 6))

for pays in top5_pays:
    donnees_pays = evolution_top5[evolution_top5['Zone'] == pays]
    plt.plot(donnees_pays['Année'], donnees_pays['Valeur'], marker='o', label=pays)

plt.title("Évolution de l'aide alimentaire (Top 5 pays, 2013-2017)")
plt.xlabel("Année")
plt.ylabel("Aide alimentaire (en kg)")

# Force l'axe des x à afficher uniquement les années entières de 2013 à 2017
plt.xticks([2013, 2014, 2015, 2016, 2017])

plt.legend()
plt.grid(True)
plt.show()
```
